# HR ANALYTICS PROJECT 

```python
import pandas as pd

data = {
    "Employee_ID": [101,102,103,104,105,106,107,108],
    "Employee_Name": ["Amit","Rahul","Priya","Sneha","Rohan","Neha","Karan","Pooja"],
    "Department": ["HR","Sales","Finance","HR","Sales","Finance","IT","IT"],
    "Salary": [40000,55000,60000,45000,70000,65000,80000,75000]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("TotalEmployees:", df["Employee_ID"].count())
```

```python
print("Average Salary:", df["Salary"]. mean())
```

```python
print("Highes Salary:", df["Salary"].max())
```

```python
print("Lowest Salary:", df["Salary"].min())
```

```python
def_count = df.groupby("Department")["Employee_ID"].count()

print(def_count)
```

```python
avg_salary = df.groupby("Department")["Salary"].mean()

print(avg_salary)
```

```python
print(avg_salary.idxmax())
```

```python
import matplotlib.pyplot as plt

avg_salary.plot(kind="bar")

plt.title("Average Salary by Department")

plt.xlabel("Department")

plt.ylabel("Salary")

plt.show()
```

# COMPLETE PROJECT CODE

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Employee_ID": [101,102,103,104,105,106,107,108],
    "Employee_Name": ["Amit","Rahul","Priya","Sneha","Rohan","Neha","Karan","Pooja"],
    "Department": ["HR","Sales","Finance","HR","Sales","Finance","IT","IT"],
    "Salary": [40000,55000,60000,45000,70000,65000,80000,75000]
}

df = pd.DataFrame(data)

print("Total Employees:", df["Employee_ID"].count())

print("Average Salary:", df["Salary"].mean())

print("Highest Salary:", df["Salary"].max())

print("Lowest Salary:", df["Salary"].min())

dept_count = df.groupby("Department")["Employee_ID"].count()

print(dept_count)

avg_salary = df.groupby("Department")["Salary"].mean()

print(avg_salary)

print("Highest Paid Department:", avg_salary.idxmax())

avg_salary.plot(kind="bar")

plt.title("Average Salary by Department")

plt.xlabel("Department")

plt.ylabel("Salary")

plt.show()
```
